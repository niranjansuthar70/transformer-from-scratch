```python
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, seq_length):
        super(PositionalEncoding, self).__init__()
        self.d_model = d_model
        self.seq_length = seq_length

    def forward(self):
        even_indices = torch.arange(0, self.d_model, 2).float()
        denominator = torch.pow(10000, (even_indices / self.d_model)).float()
        positions = torch.arange(self.seq_length).float().reshape(self.seq_length, 1)
        pe_even = torch.sin(positions / denominator)
        pe_odd = torch.cos(positions / denominator)
        stacked=torch.stack([pe_even, pe_odd], dim=2)
        pe=torch.flatten(stacked, start_dim=1, end_dim=2)
        return pe
```

```python
pe=PositionalEncoding(d_model=8 , seq_length=10)
pe.forward()
```
